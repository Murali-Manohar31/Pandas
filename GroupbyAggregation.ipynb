```python
import numpy as np
import pandas as pd 
```

```python
data={
    "Category": ["A", "B", "A", "B", "A", "B", "A", "B"],
    "Store": ["S1", "S1", "S2", "S2", "S1", "S2", "S2", "S1"],
    "Sales": [100, 200, 150, 250, 120, 180, 200, 300],
    "Quantity": [10, 15, 12, 18, 8, 20, 15, 25],
    "date":pd.date_range('2026-01-01',periods=8)
}
df=pd.DataFrame(data)
```

```python
df
```

```python
#group by catagory and sum the sales

abc=df.groupby('Category')['Sales'].sum()
abc
```

```python
#group by store calculate the sum of sales
bc=df.groupby('Store')['Sales'].sum()
bc
```

```python
#group by multiple coloumn 
#group by Catagory and Store

cd=df.groupby(['Category','Store'])['Sales'].sum()
cd
```
