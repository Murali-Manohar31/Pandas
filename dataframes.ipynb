Creating a Dataframe 

A Pandas DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns). Think of it as a spreadsheet or a SQL table inside Python, where each column represents a Pandas Series object

```python
data={
    'Name':['Murali','Devil','Loki','Spidey'],
    'Age':[21,22,25,28],
    'city':['bbsr','bombai','ohio','asgard'],
    'salary':[20000,25000,15000,16000]
    
}
df=pd.DataFrame(data)
```
