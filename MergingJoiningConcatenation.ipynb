```python
import numpy as np
import pandas as pd
```

Merging 2 data frames

```python
employees=pd.DataFrame({
    'employee_id':[1,2,3,4,5],
    'name':['John','Natha','Thor','Alice','Lee'],
    'department':['wwe','Cse','mcu','borderland','HR']
})

#dataframe 2
salaries=pd.DataFrame({
    'employee_id':[1,2,3,6,7],
    'salary':[100000,50000,12000,6900,54100],
    'bonus':[200,2000,1200,10,690]
    
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries)
```

```python
pd.merge(employees,salaries, on='employee_id',how="inner")
```

```python
pd.merge(employees,salaries,on='employee_id',how="outer")
```

```python
pd.merge(employees,salaries, on='employee_id',how="left")        
```

```python
pd.merge(employees,salaries, on='employee_id',how="right")        
```

Concatenation of dataframes


```python
df1=pd.DataFrame({
    'A':['A0','B0','C0'],
    'B':['A1','B1','C1'],
    'C':['A2','B2','C2']
})
df2=pd.DataFrame({
    'A':['A3','B3','C3'],
    'B':['A4','B4','C4'],
    'C':['A5','B5','C5']
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],axis=1)#for row concatenation
```
